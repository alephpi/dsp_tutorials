```python
# notebook debugging preamble
%load_ext autoreload
%autoreload 2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def wave(freq, phase, t):
    return np.cos(2 * np.pi * freq * t + phase)
```

```python
import ipywidgets as widgets
from ipywidgets import interact


def plot_signal_and_spectrum(sampling_duration):
    sampling_rate = 10
    N = sampling_rate * sampling_duration
    t = np.linspace(0, sampling_duration, N, endpoint=False)
    signal = wave(1, 0, t)
    spectrum = np.fft.fft(signal)[: N // 2]
    frequencies = np.fft.fftfreq(len(signal), 1 / sampling_rate)[: N // 2]

    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(t, signal)
    plt.title("Signal")
    plt.xlabel("Time [s]")
    plt.ylabel("Amplitude")

    plt.subplot(1, 2, 2)
    plt.plot(frequencies, np.abs(spectrum))
    plt.title("Spectrum")
    plt.xlabel("Frequency [Hz]")
    plt.ylabel("Magnitude")
    plt.tight_layout()
    plt.show()


interact(
    plot_signal_and_spectrum,
    sampling_duration=widgets.IntSlider(min=1, max=20, step=1, value=sampling_duration),
)
```
