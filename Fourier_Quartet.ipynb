```python
import numpy as np
```

```python
def wave(freq, phase, t):
    return np.cos(2 * np.pi * freq * t + phase)
```

```python
freq1 = 1
freq2 = np.pi
```

```python
t = np.linspace(-10 * np.pi, 10 * np.pi, 1000)
y = wave(freq1, 0, t) + wave(freq2, 0, t)
```

```python
f = np.fft.fft(y)
```
