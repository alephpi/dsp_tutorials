```python
# notebook debugging preamble
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def impulse_train(K, omega_0):
    omegas = np.arange(-K, K + 1) * omega_0
    t = np.linspace(-10, 10, 100001)
    # train = np.exp(1j * 2*np.pi * omegas * t[:, np.newaxis]).sum(axis=1)
    train = 1 + 2 * np.cos(2 * np.pi * omegas * t[:, np.newaxis]).sum(axis=1)
    return t, train
```

```python
plt.plot(*impulse_train(100, 1))
```
